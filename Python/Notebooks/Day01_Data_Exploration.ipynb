```python
#Total Revenue


query = """
SELECT SUM(Delivery_Fee) AS Total_Revenue
FROM deliveries
"""

pd.read_sql(query, connection)
```

```python
#Revenue by Driver

query = """
SELECT Driver,
       SUM(Delivery_Fee) AS Revenue
FROM deliveries
GROUP BY Driver
ORDER BY Revenue DESC
"""

pd.read_sql(query, connection)
```

```python
#Total Profit by Driver

query = """
SELECT Driver,
       SUM(Profit) AS Total_Profit
FROM deliveries
GROUP BY Driver
ORDER BY Total_Profit DESC
"""

pd.read_sql(query, connection)
```
