```python
query = """
SELECT
    Delivery_ID,
    Driver,
    Profit,

    SUM(Profit)
    OVER(PARTITION BY Driver)
    AS Driver_Total_Profit

FROM deliveries;
"""

pd.read_sql(query, connection)
```

```python
#Write a query that classifies delivery times:

#Less than 45 minutes → "Fast"
#Between 45 and 90 minutes → "Normal"
#More than 90 minutes → "Slow"



query = """
SELECT
    Delivery_ID,
    Delivery_Time_Min,

    CASE

        WHEN Delivery_Time_Min >= 90 THEN 'Slow'

        WHEN Delivery_Time_Min >= 45 THEN 'Normal'

        ELSE 'Fast'

    END AS Delivery_Time_Min_Category

FROM deliveries;
"""

pd.read_sql(query, connection)

```

```python
#Customer with the highest total profit and their rank among all customers based on profit.


query = """
SELECT
    Customer_Name,
    SUM(Profit) AS Total_Profit,
    RANK() OVER(ORDER BY SUM(Profit) DESC) AS Profit_Rank
FROM deliveries
GROUP BY Customer_Name;
"""

pd.read_sql(query, connection)
```

```python
query = """
WITH Route_Stats AS
(
    SELECT
        Route,
        AVG(Profit) AS Route_Profit
    FROM deliveries
    GROUP BY Route
)

SELECT
    Route,
    Route_Profit,

    CASE
        WHEN Route_Profit > 2500 THEN 'Excellent'
        WHEN Route_Profit >= 1500 THEN 'Good'
        ELSE 'Poor'
    END AS Performance

FROM Route_Stats

ORDER BY Route_Profit DESC;
"""

pd.read_sql(query, connection)
```
